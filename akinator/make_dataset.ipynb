```python
EXTRA_FEATURES_V2 = ["produces_milk", "has_long_ears"]
EXTRA_VALUES_V2 = {
    "Cow":    [1, 0],
    "Horse":  [0, 0],
    "Pig":    [0, 0],
    "Rabbit": [0, 1],
}

features = FEATURES.copy()
rows = [r.copy() for r in ROWS]

if VERSION == "v2":
    features += EXTRA_FEATURES_V2
    for r in rows:
        extras = EXTRA_VALUES_V2.get(r[0], [0] * len(EXTRA_FEATURES_V2))
        r.extend(extras)

## Make Dataframe
data = pd.DataFrame(rows, columns=["name"] + features)

# SAVE
data.to_csv(f"data/entities_{VERSION}.csv", index=False)
```

```python
# Feature balance check — aim for 30–70% positive per column.
# Columns outside that range are weak splitters (carry less info gain).

balance = (data[FEATURES].sum() / len(data) * 100).round(1).sort_values()
weak = balance[(balance < 30) | (balance > 70)]
print("Feature positive rate (%):\n")
print(balance.to_string())
print(f"\nFeatures outside 30–70% range (weaker splitters): {list(weak.index)}")
```
